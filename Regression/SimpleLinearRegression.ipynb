***LINEAR REGRESSION***

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/Users/ishaangupta/Downloads/GIT HUB FILES/placement.csv")
```

```python
class LinearRegresion:
    def __init__(self):
        self.m=None
        self.b=None
    def fit(self, x_train, y_train):
        num = 0
        den = 0

        for i in range(x_train.shape[0]):
            num = num + ((x_train.iloc[i] - x_train.mean()) *
                        (y_train.iloc[i] - y_train.mean()))

            den = den + ((x_train.iloc[i] - x_train.mean()) *
                        (x_train.iloc[i] - x_train.mean()))

        self.m = num / den
        self.b = y_train.mean() - (self.m * x_train.mean())
    def predict(self,x_test):
       
        print(self.m * x_test + self.b)
```

```python
x=df.iloc[:,0]
y=df.iloc[:,1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=2)
```

```python
lr=LinearRegresion()
lr.fit(x_train,y_train)
lr.predict(x_test.iloc[0])
```
