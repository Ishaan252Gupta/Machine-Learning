```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/Users/ishaangupta/Downloads/GIT HUB FILES/placement.csv")
df.head()
```

```python
plt.scatter(x=df["cgpa"],y=df["package"])
plt.xlabel("CGPA")
plt.ylabel("PACKAGE")
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df[["cgpa"]],df[["package"]],test_size=.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
plt.scatter(df["cgpa"],df["package"])
plt.plot(x_train,lr.predict(x_train),color="red")
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
y=lr.predict(x_test)
print("MSE",mean_squared_error(y_test,y))
print("MAE",mean_absolute_error(y_test,y))
```
