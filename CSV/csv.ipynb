USING REQUESTS


```python
import requests
import pandas as pd
from io import StringIO

url = "https://raw.githubusercontent.com/umpirsky/country-list/master/data/af_NA/country.csv"

req = requests.get(url)
data = StringIO(req.text)

df = pd.read_csv(data)

print(df)
```
